```python
# Melt Vs Pivot

df4 = pd.DataFrame({
    "Country": ["USA", "USA", "India", "India"],
    "Year" : [2020, 2021, 2020, 2021],
    "Sales" : [100, 120, 90, 110],
    "Profit" : [20, 25, 18, 22],


})
```

```python
df4
```

```python
# Melt
melted_df = df4.melt(
    id_vars = ["Country", "Year"],
    value_vars=["Sales","Profit"],
    var_name = "Metrics",
    value_name = "Values"
)
melted_df
```

```python
# Pivot

original = melted_df.pivot(
    index=["Country","Year"],
    columns=["Metrics"],
    values=["Values"]
)

original
```

```python
df1 = pd.DataFrame({
    "Emp_ID" : [ 101,102,103],
    "Name" : ["John", "Jane", "Alex"]
})

df2 = pd.DataFrame({
    "Emp_ID" : [101, 102, 104],
    "Salary" : [50000, 60000, 70000]

})

df1
df2
```

```python
# Merge : Most commanly used </> python
result = pd.merge(df1,df2, on="Emp_ID")
print(result)
```

```python
#INNER JOIN : Return only matching records from both dataFrame.
pd.merge(
    df1,
    df2,
    on ="Emp_ID",
    how = "inner"
)
```

```python
#LEFT JOIN : Keeps all recoreds from the left DataFrame.

pd.merge(
    df1,
  df2,
    on = "Emp_ID",
    how = "left"
)
```

```python
# RIGHT JOIN : Keeps all records from right DataFrame.

pd.merge(
    df1,
    df2,
    on = "Emp_ID",
    how = "right"
)
```

```python
#Outer JOIN : Keeps all records from both DataFrames.

pd.merge(
    df1,
    df2,
    on = "Emp_ID",
    how = "outer"
)
```

```python
# Merge using different column names.

df1 = pd.DataFrame({
  "Employee_ID" : [101,102,103],
  "Name"        : ["John","Jane","Alex"]
})

df2 = pd.DataFrame({
    "Emp_ID" : [101,102,104],
    "Salary" : [50000,60000,70000]

})

result = pd.merge(
    df1,
    df2,
    left_on="Employee_ID",
    right_on="Emp_ID",
    how = "left"
)
result
```

```python
# Concat() : Don't confuse it with merge.
# Concat() is generally used to stack DataFrame.
# Vertical :

result_vertical = pd.concat([df1, df2])
result_vertical

# Horizontal:

result_horizontal = pd.concat([df1,df2], axis=1)
result_horizontal
```
