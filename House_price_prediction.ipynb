```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
# X= square footage; y=price

X = [[1000],[1200],[1500],[1800],[2200]]
Y = [40, 48,60,72,88] #price inlakhs

X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)

predicted_price = model.predict([[1600]])
print(predicted_price[0])
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

data = pd.DataFrame({
    "day_of_week": [0, 0, 1, 4, 4],  # Monday=0
    "hour":        [9, 10, 9, 17, 18],
    "is_holiday":  [0, 0, 0, 0, 0],
    "calls":       [120, 160, 135, 80, 60],
    "agents_needed": [15, 20, 17, 10, 8]
})

# Features known when planning the schedule
X = data[["day_of_week", "hour", "is_holiday", "calls"]]
y = data["agents_needed"]

model = RandomForestRegressor(random_state=42)
model.fit(X, y)

# Forecast staffing need: Monday, 10 AM, normal day, ~170 calls expected
prediction = model.predict([[0, 10, 0, 170]])
print(f"Recommended agents: {round(prediction[0])}")
```
