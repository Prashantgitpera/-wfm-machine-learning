```python
import matplotlib.pyplot as plt

# Check available styls

print(plt.style.available)

# Apply style

plt.style.use("ggplot")

x = [1,2,3,4,5]
y = [10,20,15,25,30]

plt.plot(

x,y,
linewidth =2,
marker="o",
label="sales"
)
plt.title("Sales performance")
plt.xlabel("Month")
plt.ylabel("sales")

#Add grid

plt.grid(True, linestyle="--", linewidth=0.5)

# Show legend

plt.legend()

#Adjust layout

plt.tight_layout()

# Save figure

plt.savefig("sales_performance.png")

# Display

plt.show()
```
