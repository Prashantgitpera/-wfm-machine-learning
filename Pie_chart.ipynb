Pie chart _ max 5- 6 catagories.

```python
import matplotlib.pyplot as plt

labels = ['Apple', 'Banana', 'Orange', 'Grapes']
sizes = [35, 25, 20, 20]
explode = (0.08, 0, 0, 0)  # Explode the first slice

plt.pie(
    sizes,
    labels=labels,
    explode=explode,
    autopct='%1.1f%%',
    shadow=True,
    startangle=140
)

plt.title("Fruit Distribution")
plt.axis('equal')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
labels = ['Python', 'Java', 'C++', 'JavaScript']
sizes = [40, 25, 20, 15]

# Create pie chart
plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',   # Show percentages
    startangle=90        # Rotate chart
)

plt.title("Programming Language Popularity")
plt.axis('equal')  # Makes the pie chart circular
plt.show()
```

Example of wedgeprops

```python
import matplotlib.pyplot as plt

labels = ['Python', 'Java', 'C++', 'JavaScript']
sizes = [40, 25, 20, 15]

fig, ax = plt.subplots()
ax.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',
    wedgeprops={'edgecolor': 'black'}
)

fig.patch.set_facecolor('#f0f0f0') # Change background color using a hex code

plt.title("Programming Language Usage")
plt.show()
```

Donut Pie chart

```python
import matplotlib.pyplot as plt

labels = ['Python', 'Java', 'C++', 'JavaScript']
sizes = [40, 25, 20, 15]

plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',
    wedgeprops={
        'width': 0.5,
        'edgecolor': 'white'
    }
)

plt.title("Programming Language Usage")
plt.show()
```

Market share of WFM tools.

```python
import matplotlib.pyplot as plt

tools = [
    'NICE IEX WFM',
    'Verint WFM',
    'Genesys WFM',
    'Alvaria Workforce',
    'injixo'
]

market_share = [36.9, 30.2, 7.8, 7.8, 5.0]

plt.pie(
    market_share,
    labels=tools,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={
        'edgecolor': 'white',
        'linewidth': 2
    }
)

plt.title('Top 5 Global Contact Center WFM Tools')
plt.axis('equal')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Data
tools = [
    'NICE IEX WFM',
    'Verint WFM',
    'Genesys WFM',
    'Alvaria Workforce',
    'injixo'
]

market_share = [36.9, 30.2, 7.8, 7.8, 5.0]

# Seaborn styling
sns.set_theme(style="dark")

# Pie chart
plt.pie(
    market_share,
    labels=tools, # Labels for the slices
    autopct='%1.1f%%',
    startangle=90,
    pctdistance=0.8, # Adjust this value to move percentages inside the slices
    wedgeprops={
        'edgecolor': 'white',
        'linewidth': 1
    }
)

plt.title('Top 5 Global WFM Tools – Market Share')
plt.axis('equal')
plt.legend(tools, title="WFM Tools", loc="center left", bbox_to_anchor=(1, -0.3, 0.5, 1)) # Move legend further below
plt.show()
```
