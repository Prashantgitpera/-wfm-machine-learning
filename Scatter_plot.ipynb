Offered Calls vs AHT or Offered Calls vs Required HC.

```python
import matplotlib.ticker as mticker

offered_calls = [
    82500, 79800, 86200, 88900,
    91500, 94200, 97800, 95600,
    99400, 103200, 108500, 115800]

aht = [
    4.2, 4.0, 4.4, 4.5,
    4.6, 4.7, 4.8, 4.5,
    4.9, 5.0, 5.1, 5.3
]

#Change color

colors = ["#fac778" if call < 90000 else "#e46c59" for call in offered_calls]

# Resize
sizes = [val * 12 for val in aht] # Scale AHT values to increase marker size

# Create the scatter plot
plt.figure(figsize=(8, 4))
plt.scatter(offered_calls, aht, color=colors, s=sizes, marker="x", alpha = 1)

plt.title("offered_calls vs AHT")
plt.xlabel("offered_calls")
plt.ylabel("AHT")

#change offered calls format

def thousands_formatter(x, pos):
    return f'{int(x/1000)}k'

formatter = mticker.FuncFormatter(thousands_formatter)
plt.gca().xaxis.set_major_formatter(formatter)
plt.grid()
plt.xlim (min(offered_calls)-5000, max(offered_calls)+15000)

plt.show()
```

```python
requireede_hc = [
    32, 30, 35, 36, 38, 40, 42, 39, 44, 46, 48, 52
]
  # Color map

plt.scatter(offered_calls, requireede_hc, cmap="OrRd", c=offered_calls)

def thousands_formatter(x, pos):
    return f'{int(x/1000)}k'

formatter = mticker.FuncFormatter(thousands_formatter)
plt.gca().xaxis.set_major_formatter(formatter)
plt.xlim (min(offered_calls)-5000, max(offered_calls)+15000)
plt.grid()
```
