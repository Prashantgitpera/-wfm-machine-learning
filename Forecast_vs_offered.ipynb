WFM Offered calls volume plot using Matplotlib.

```python


from matplotlib import colors
import matplotlib.pyplot as plt

# Check available styls

print(plt.style.available)

# Apply style

plt.style.use("ggplot")

x=["Jan","Feb","Mar","Apr","May","June","July","Aug","Sep","Oct","Nov","Dec"]
y=[82500,79800,86200,88900,91500,94200,97800,95600,99400,103200,108500,115800]
y_forecast=[81000,81500,84500,90000,92500,93000,96500,98000,101000,105000,107000,113000]

plt.plot(
         x,y_forecast,
         linewidth=2,
         color="#00A693",
         marker="x",
         label="2025 forecast volume"
)


plt.plot(

         x,y,
         linewidth =2,
         color= "#494e84",
         marker="o",
         label="2025 offered calls"
)

plt.tick_params(axis="x", rotation=45)
plt.title("Forcast vs Offered")
plt.xlabel("Month")
plt.ylabel("offered calls")

#add grid

plt.grid(True, linestyle="--", linewidth=0.5)

# Show legend

plt.legend(loc="upper left")
plt.tight_layout()

```

```python
difference = [actual - forecast for actual, forecast in zip(y, y_forecast)]

# Display the differences
for month, diff in zip(x, difference):
    print(f"Difference for {month}: {diff}")

# Optionally, you can also plot the difference
plt.figure(figsize=(10, 6))
plt.plot(x, difference, marker='o', linestyle='-', color='#da9100')
plt.title('Difference Between Actual and Forecast Calls')
plt.xlabel('Month')
plt.ylabel('Difference (Actual - Forecast)')
plt.grid(True, linestyle='--', linewidth=0.5)
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('difference_plot.png') # Save the plot
plt.show()

```
