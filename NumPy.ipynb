```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

mean = np.mean(arr)
std_dev = np.std(arr)

normalized = (arr - mean) / std_dev

print("Mean:", mean)
print("Standard Deviation:", std_dev)
print("Normalized Array:", normalized)
```

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

# Population standard deviation (default)
std_pop = np.std(arr)

# Sample standard deviation
std_sample = np.std(arr, ddof=1)

print(std_pop)
print(std_sample)
```
