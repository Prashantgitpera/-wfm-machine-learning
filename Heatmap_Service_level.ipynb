```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# WFM service-level data: percentage of requests answered within SLA
data = {
    "hour": [
        "09:00", "10:00", "11:00", "12:00", "13:00",
        "14:00", "15:00", "16:00", "17:00"
    ],
    "Monday":    [88, 84, 79, 76, 82, 85, 80, 78, 86],
    "Tuesday":   [90, 86, 81, 77, 80, 83, 79, 76, 84],
    "Wednesday": [87, 83, 78, 74, 79, 82, 77, 75, 83],
    "Thursday":  [91, 88, 84, 80, 83, 86, 82, 79, 87],
    "Friday":    [85, 80, 75, 71, 76, 79, 74, 72, 80]
}

df = pd.DataFrame(data)

# Convert into heatmap-friendly format
heatmap_data = df.set_index("hour")

plt.figure(figsize=(11, 6))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".0f",
    cmap="coolwarm",
    vmin=60,
    vmax=100,
    linewidths=0.5,
    cbar_kws={"label": "Service Level (%)"}
)

plt.title("WFM Service Level Heatmap by Day and Hour")
plt.xlabel("Day")
plt.ylabel("Hour")
plt.show()
```
