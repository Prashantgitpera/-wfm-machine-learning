```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn  as sns
```

Create database

```python
df = pd.DataFrame({
    'Month' : ["Jan","Feb","Mar","Apr","May"],
    "Sales" : [ 120, 150, 180, 140, 200],
    "Order" : [20, 25, 30, 22, 35]
})
df
```

Line chart Trend

```python
plt.plot(df["Month"], df['Sales'], marker= 'o')
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

Bar Chat

```python
plt.bar(df["Month"], df['Sales'])
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

Histogram-Distribution

```python
plt.hist(df["Sales"], bins = 5)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

Scatter plot relationship

```python
plt.scatter(df["Month"], df['Sales'])
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

Box-plot

```python
plt.boxplot(df["Sales"])
plt.title("Sales Distribution")
plt.ylabel("Sales")

plt.show()
```

For a Data Analyst, a good learning sequence is:
Pandas visualization → Matplotlib → Seaborn → Plotly → Dashboards.

```python

df1 = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May'],
    'Sales': [100, 120, 150, 140, 180]
})

df2 = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May'],
    'Sales': [110, 130, 145, 160, 200]
})

plt.plot(
    df1['Month'],
    df1['Sales'],
    marker='o',
    label='2025 Sales'
)

plt.plot(
    df2['Month'],
    df2['Sales'],
    marker='o',
    label='2026 Sales'
)

plt.title('2025 vs 2026 Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.legend()

plt.show()
```
