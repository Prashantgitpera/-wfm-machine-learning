```python
# Linear regression
import numpy as np

# Historical WFM data: hour of day and incoming calls
hours = np.array([9, 10, 11, 12], dtype=float)
calls = np.array([100, 130, 160, 190], dtype=float)

m, b = 0.0, 0.0
learning_rate = 0.01
iterations = 1000
n = len(hours)

for _ in range(iterations):
    predictions = m * hours + b
    errors = predictions - calls

    # Gradient calculation
    dm = (2 / n) * np.sum(hours * errors)
    db = (2 / n) * np.sum(errors)

    # Gradient-descent update
    m -= learning_rate * dm
    b -= learning_rate * db

forecast_1pm = m * 13 + b
print(f"Forecasted calls at 1 PM: {forecast_1pm:.0f}")
```

```python
import numpy as np

hours = np.array([9, 10, 11, 12], dtype=float)
calls = np.array([100, 130, 160, 190], dtype=float)

# Scale hours so gradient descent is stable
hour_mean = hours.mean()
hour_std = hours.std()
x = (hours - hour_mean) / hour_std

m, b = 0.0, 0.0
learning_rate = 0.1
iterations = 1000
n = len(x)

for _ in range(iterations):
    predictions = m * x + b
    errors = predictions - calls

    dm = (2 / n) * np.sum(x * errors)
    db = (2 / n) * np.sum(errors)

    m -= learning_rate * dm
    b -= learning_rate * db

# Forecast calls at 1 PM (13:00)
hour_to_predict = 13
scaled_hour = (hour_to_predict - hour_mean) / hour_std
forecast_1pm = m * scaled_hour + b

print(f"Forecasted calls at 1 PM: {forecast_1pm:.0f}")
```

```python
import pandas as pd
import seaborn as sns
df= pd.read_csv("/content/drive/My Drive/Learning/Prime 2.0 AI ML (APNA College)/Machine learning/Supervised machine learning/insurance.csv")
df
```

```python
# Assigning smoker yes-1, no-0
df.loc[df["smoker"]=="yes","smoker"]=1
df.loc[df["smoker"]=="no","smoker"]=0
df
```

```python
from ipywidgets import Label
# scatter plot
df.head()
sns.scatterplot(x=df["bmi"], y=df["charges"], hue=df["smoker"])

```

```python
 X = df.drop(columns = ["charges", "region"])
 Y = df["charges"]
X.loc[df["sex"]=="female","sex"]=1
X.loc[df["sex"]=="male","sex"]=0
 X.head()
```

```python
# Train test split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42)
```

```python
x_test.head()
```

```python
y_test.head()
```

```python
# Train model

from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train, y_train)
```

```python
# Predict values

y_pred = model.predict(x_test)
y_pred
```

```python
# Evalute R-square

from sklearn.metrics import mean_squared_error, r2_score
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
r2
print("MSE: ", mse)
print("R2: ", r2)
```

```python
# Evalute Adjusted R-square

n = x_test.shape[0]
p=x_test.shape[1]
adj_r2 = 1 - (1 - r2) * ((n - 1) / (n - p - 1))
print("Adjusted R2: ", adj_r2)
```
