Bar Chart

```python
import matplotlib.pyplot as plt
```

```python
months = ["Jan","Feb","Mar","Apr"]
calls = [82500, 79800, 86200, 88900]

plt.bar(months,calls)
plt.title("Monthly offered calls - 2025")
plt.xlabel("Months")
plt.ylabel("offered Calls")
plt.show()
```

Customize the Bar chart.

```python
plt.bar(
    months,
    calls,
    width=0.6,
    alpha=0.8,
    label="Offered calls"
)

plt.title("WFM Offered Calls-2025")
plt.xlabel("Months")
plt.ylabel("Offered Calls")
plt.legend()
plt.grid(axis="y")
plt.tick_params(axis="x",rotation=45)
plt.yticks(
    [80000, 60000, 40000, 20000],
    ["80k","60k","40k","20k"]
)
plt.show()
```

Forecast vs Actual Bar chart

```python
months = ["Jan", "Feb","Mar","Apr","May","Jun"]
forecast = [81000, 81500, 84500, 90000, 92500, 93000]
actual = [82500, 79800, 86200, 88900, 91500, 94200]

x = range(len(months))
width = 0.35

plt.bar(
    [i - width/2 for i in x],
    forecast,
    width=width,
    label ="Forecast"

)

plt.bar(
    [i + width/2 for i in x],
    actual,
    width=width,
    label="Actual"
)

plt.xticks(x,months)
plt.title("Forecast vs Actual Offered Calls")
plt.xlabel("Month")
plt.ylabel("Calls")
plt.legend()
plt.grid(axis="y")
plt.yticks(
    [80000, 60000, 40000, 20000],
    ["80k","60k","40k","20k"]
)
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()

```

Horizontal Bar chart.

```python
months = ["Jan", "Feb","Mar","Apr","May","Jun"]
forecast = [81000, 81500, 84500, 90000, 92500, 93000]

plt.barh(months,forecast, color="#da9100")

plt.xlabel("Offered Calls")
plt.ylabel("Months")
plt.title("Monthly Offered Calls")
plt.yticks(rotation=45)
plt.xticks(
    [80000, 60000, 40000, 20000],
    ["80k","60k","40k","20k"]
)
plt.tight_layout()
# Add text labels on the bars
for index, value in enumerate(forecast):
    plt.text(value, index, str(value), va='center', ha='left')
plt.xlim(0,max(forecast)+15000)
plt.show()
```
