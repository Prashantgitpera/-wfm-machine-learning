```python
import pandas as pd

data = pd.DataFrame({
    "call_type": ["Billing", "Technical", "Sales", "Billing"],
    "calls": [120, 90, 75, 140]
})

encoded_data = pd.get_dummies(data, columns=["call_type"], dtype=int)

encoded_data
```
